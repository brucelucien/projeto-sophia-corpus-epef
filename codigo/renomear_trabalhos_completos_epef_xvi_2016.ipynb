# Renomear trabalhos completos do EPEF XVI (2016)

```python
import os
```

```python
diretorio = '../dados/epef_xvi_2016/'
nomes_arquivos = os.listdir(diretorio)
for nome_arquivo in nomes_arquivos:    
    sequencial = nome_arquivo.removesuffix('.pdf')
    novo_nome_arquivo = '2016-' + str(sequencial).rjust(3, '0') + '.pdf'
    os.rename(diretorio + nome_arquivo, diretorio + novo_nome_arquivo)
```
