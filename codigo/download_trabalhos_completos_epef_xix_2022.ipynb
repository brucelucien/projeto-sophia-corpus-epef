# Download dos trabalhos completos do EPEF XIX (2022)

```python
from bs4 import BeautifulSoup
import requests
```

```python
html = requests.get('https://sec.sbfisica.org.br/eventos/epef/xix/atas/TrabalhosPorArea.asp')
bs = BeautifulSoup(html.text)
links = bs.find('div', class_='container').find_all('a')
ULTIMO_PDF_A_CONSIDERAR = 'T0216-1.pdf'
terminar = False
cont = 0
while cont < len(links) and not terminar:
    link = links[cont]
    endereco_pdf = link.get_attribute_list('href').pop().strip()
    nome_pdf =  endereco_pdf.split('/')[-1]
    pdf = requests.get(endereco_pdf)
    prefixo_identificador = '2022-' + str(cont).rjust(3, '0')
    with(open(f'../dados/epef_xix_2022/{prefixo_identificador}-{nome_pdf}', 'wb')) as arquivo:
        arquivo.write(pdf.content)
    cont += 1
    terminar = nome_pdf.strip().upper() == ULTIMO_PDF_A_CONSIDERAR.strip().upper()    
    
```
